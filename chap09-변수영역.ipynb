## 지역변수와 전역변수

### 지역변수, 전역변수

````python
%%writefile func.c
#include "myheader.h"
extern int gVal;

void A()
{
	int value = 20;
	gVal++;
}
````

````python
%%writefile myheader.h
void A();
````

````python
%%writefile ex9_1.c
#include <stdio.h>
#include "myheader.h"
int gVal = 1;

int main(void)
{
	int value = 10;
	printf("지역변수 value 값은 %d이다.\n", value);
	printf("전역변수 gVal 값은 %d이다.\n", gVal);
	A();
	printf("함수 A() 호출 후 \n");
	printf("지역변수 value 값은 %d이다.\n", value);
	printf("전역변수 gVal 값은 %d이다.\n", gVal);
	return 0;
}   
````

파일이 총 3개로 나뉘어진 프로젝트 이다. 따라서 각 소스파일(func.c, ex9_1.c)를 모두 컴파일하고, 링크를 수행한다.

````python
%%shell
gcc ex9_1.c func.c -o ex9_1
./ex9_1
````

### 변수의 우선순위

````python
%%writefile ex9_2.c
#include <stdio.h>
void A();
int Number = 1;

int main(void)
{
	int Number = 10;
	printf("지역변수 Number 값은 %d이다.\n", Number);
	{
		int Number ;
		Number = 20;
		printf("블록 내 지역변수 Number 값은 %d이다.\n", Number);
	}

	printf("전역변수 Number 값은 %d이다.\n", Number);
	A();

	return 0;
}

void A()
{
	Number++;
	printf("전역변수 Number 값은 %d이다.\n", Number);
}
````

````python
%%shell
gcc ex9_2.c -o ex9_2
./ex9_2
````

## 정적변수

### 내부 정적변수

````python
%%writefile ex9_3.c
#include <stdio.h>
void CallNumber();

int main(void)
{
	int i;
	for(i=0; i<5; i++)
		CallNumber();

	return 0;
}

void CallNumber()
{
	static int count = 1;
	int number = 1;
	printf("%d번째 호출되었습니다. \n", count++);
	printf("Number 값은 %d이다. \n", number++);
}
````

````python
%%shell
gcc ex9_1.c -o ex9_1
./ex9_1
````

### 레지스터 변수

````python
%%writefile ex9_4.c
#include <stdio.h>

int main(void)
{
	register int count = 1;
	int result = 0;

	for(count = 1 ; count <= 5; count++)
		result += count;   

	printf("1부터 5까지의 합은 %d이다.\n", result);
	return 0;
}
````

````python
%%shell
gcc ex9_4.c -o ex9_4
./ex9_4
````
