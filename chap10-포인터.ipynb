## 포인터의 기본

포인터는 C 언어의 대표적인 특징 중 하나로, 변수나 데이터가 저장된 메모리의 주소를 저장하는 자료형이다. 포인터를 사용하면 변수 이름을 직접 사용하지 않고도 해당 메모리 공간에 접근할 수 있어 프로그램을 효율적으로 작성할 수 있다. 그러나 포인터는 이해하기 어려운 개념이며, 잘못 사용할 경우 메모리 오류가 발생해 프로그램 실행에 심각한 영향을 줄 수 있다.

### 포인터 선언, 포인터를 이용한 일반 변수로의 접근

```python
%%writefile ex10_1.c
#include <stdio.h>

int main(void)
{
	int i = 10;   
	int *ptr = &i;

	printf("ptr의 주소 값: %p\n", &ptr);
	printf("i의 주소 값: %p\n", &i);
	printf("ptr이 가리키는 주소 값: %p\n\n", ptr);

	printf("i의 값: %d\n", i);
	printf("ptr이 가리키는 값: %d\n", *ptr);

	return 0;
}
```

```python
%%shell
gcc ex10_1.c -o ex10_1
./ex10_1
```

```python
%%writefile ex10_2.c
#include <stdio.h>

int main(void)
{
	int i = 10;
	int *ptr = &i;     

	*ptr = i+20;
	printf("i ∞™ : %d \n", i);

	i = i + 20;
	printf("i ∞™ : %d \n", *ptr);

	return 0;
}
```

```python
%%shell
gcc ex10_2.c -o ex10_2
./ex10_2
```

### 포인터의 유용성

ex10_3.c 와 ex10_4.c 의 결과를 비교하자

```python
%%writefile ex10_3.c
#include <stdio.h>

void Swap(int a, int b);
int main(void)
{
	int x = 10, y = 5;     
	printf("Swap() 함수 실행 전\n");
	printf("x = %d, y = %d이다.\n\n", x, y);

	Swap(x, y);
	printf("Swap() 함수 실행 후\n");
	printf("x = %d, y = %d이다.\n", x, y);

	return 0;
}

void Swap(int a, int b)
{
	int temp;
	temp = a;
	a = b;
	b = temp;
}
```

```python
%%shell
gcc ex10_3.c -o ex10_3
./ex10_3
```

```python
%%writefile ex10_4.c
#include <stdio.h>

void Swap(int* a, int* b);
int main(void)
{    
	int x = 10, y = 5;
	printf("Swap() 함수 실행 전\n");
	printf("x = %d, y = %d이다.\n\n", x, y);

	Swap(&x, &y);
	printf("Swap() 함수 실행 후\n");
	printf("x = %d, y = %d이다.\n", x, y);

	return 0;
}

void Swap(int* a, int* b)
{
	int temp;
	temp = *a;
	*a = *b;
	*b = temp;
}
```

```python
%%shell
gcc ex10_4.c -o ex10_4
./ex10_4
```

## 인자 전달 방법

- 인자를 값으로 전달 call-by-value, 인자를 주소로 전달 call-by-reference

```python
%%writefile ex10_5.c
#include <stdio.h>
void CountIncrement1(int n); // 값을 전달
void CountIncrement2(int* n); // 포인터를 전달

int main(void)
{
	int a = 10;
	printf("a의 초깃값 : %d\n", a);
	     
	CountIncrement1(a);
	printf("CountIncrement1 함수 실행 후 a의 값 : %d\n", a);

	CountIncrement2(&a);
	printf("CountIncrement2 함수 실행 후 a의 값 : %d\n", a);

	return 0;
}

void CountIncrement1(int n)
{
	n++;
}

void CountIncrement2(int* n)
{
	(*n)++;
}
```

```python
%%shell
gcc ex10_5.c -o ex10_5
./ex10_5
```

## 포인터와 배열

### 포인터와 배열의 관계

배열명은 그 자체로 해당 데이터가 있는 시작 주소를 의미하는 포인터 상수다. 배열 원소에 배열명을 초깃값으로 가진 포인터를 이용하여 접근할 수 있다. 배열명은 상수이기 때문에 증감연산자 등으로 배열명의 값을 바꿀 수 없다. 포인터는 증감연사자 등을 이용하여 값을 바꾸는 것이 가능하다.

```python
%%writefile ex10_6.c
#include <stdio.h>

int main(void)
{   
	int a[ ] = {10, 20, 30, 40, 50};
	int *p = a;
	int i;

	printf("배열명 a를 이용한 주소 표현\n");
	for(i=0; i<5; i++)
		printf("a[%d]의 주소: %p\n", i, a + i);

	printf("\n포인터 p를 이용한 주소 표현\n");
	for(i=0; i<5; i++)
		printf("a[%d]의 주소: %p\n", i, p + i);

	printf("\n포인터 p를 이용한 값 표현\n");
	for(i=0; i<5; i++)
		printf("a[%d] = %d\n", i, *(p + i));

	return 0;
}
```

```python
%%shell
gcc ex10_6.c -o ex10_6
./ex10_6
```

```python
%%writefile ex10_7.c
#include <stdio.h>

int main(void)  
{
	int a[ ] = {11, 22, 33, 44, 55};
	int *p = a;

	printf("*p의 값 = %d\n", *p);
	p++;

	printf("*(p+1)의 값 = %d\n", *p);
	p++;

	printf("*(p+2)의 값 = %d\n", *p);
	p--;

	printf("*(p+1)의 값 = %d\n", *p);

	return 0;
}
```

```python
%%shell
gcc ex10_7.c -o ex10_7
./ex10_7
```

```python
%%writefile ex10_8.c
#include <stdio.h>
int SumArray(int* pA, int Size);

int main(void)    
{
	int a[ ] = {10, 5, 15, 25, 7};
	int Sum;
	Sum = SumArray(a, 5);
	printf("배열 원소의 합 : %d\n", Sum);

	return 0;
}
int SumArray(int* pA, int Size)
{
	int result = 0, i;

	for(i=0; i<Size; i++)
		result += pA[i];

		// result += *(pA + i);
		// result += *pA++;

	return result;
}
```

## 문자 처리 함수

```python
%%writefile ex10_11.c
#include <stdio.h>

int main(void)
{
	char *pC = "c language";

	while (*pC) 
	{    
		printf("%c", *pC-32);
		pC++;
	}

	printf("\n");
	return 0;
}
```

```python
%%writefile ex10_12.c
#include <stdio.h>
#include <ctype.h>

int main(void)
{
	char *pC = "c language";

	while (*pC) 
	{   
		printf("%c", toupper(*pC++));
	}

	printf("\n");
	return 0;
}
```

### 문자열 처리 함수

```python
%%writefile ex10_13.c
#include <stdio.h>

int strleng(char *ch);
int main(void)
{
	int len;
	char str[ ] = "c language and data structure";
	len = strleng(str);
	printf("문자열의 크기는 %d다.\n", len);
	return 0;
}    

int strleng(char *ch)
{
	int index;
	for (index = 0; *(ch+index) != '\0'; index++);

	return index;
}
```
